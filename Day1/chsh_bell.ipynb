# CHSH / Bell Inequality Exercise

## Context and motivation

In Lecture 2, you saw that the Bell state

$$
|\Phi^+\rangle = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)
$$

cannot be written as a product of two single-qubit states. The two qubits are therefore **entangled**.

In this exercise, you will see one of the experimental consequences of this entanglement.

Alice and Bob each receive one qubit of the Bell state. They each choose between **two different measurement directions**, and compare their measurement results.

The strength of the correlations is summarised by a number called the **CHSH value**, $S$.

For any classical local hidden-variable theory,

$$
S \leq 2.
$$

Quantum mechanics allows entangled qubits to reach

$$
S = 2\sqrt{2} \approx 2.83.
$$

Your goal is to calculate $S$ for an entangled pair and see the difference between the quantum and classical limits.

## 1. The four measurements

Alice has two possible measurement settings:

$$
a = 0, \qquad a' = \frac{\pi}{2}
$$

Bob also has two:

$$
b = \frac{\pi}{4}, \qquad b' = \frac{3\pi}{4}.
$$

This means you need to perform **four different experiments**:

| Experiment |      Alice |         Bob |
| ---------- | ---------: | ----------: |
| 1          |      $a=0$ |   $b=\pi/4$ |
| 2          |      $a=0$ | $b'=3\pi/4$ |
| 3          | $a'=\pi/2$ |   $b=\pi/4$ |
| 4          | $a'=\pi/2$ | $b'=3\pi/4$ |

For each experiment, run your circuit many times and record how often you get each of the four possible outcomes:

$$
00,\quad 01,\quad 10,\quad 11.
$$

### How do we measure at an angle?

Your quantum computer measures in the standard computational ($Z$) basis.

To instead measure along a direction at angle $\theta$, rotate the qubit first using

```python
qc.ry(-theta, qubit)
```

and then perform the usual Z-basis measurement.

The provided `correlator(angle_a, angle_b)` function should do this for you.

## 2. Calculate the correlation

For each pair of measurement settings, calculate

$$
E = P(\text{same}) - P(\text{different}).
$$

Here:

* **same** means getting `00` or `11`
* **different** means getting `01` or `10`

Therefore,

$$
P(\text{same}) = P(00)+P(11)
$$

and

$$
P(\text{different}) = P(01)+P(10).
$$

For example, if your circuit gives:

```text
00: 460
01: 40
10: 40
11: 460
```

out of 1000 shots, then

$$
P(\text{same}) = \frac{460+460}{1000}=0.92
$$

and

$$
P(\text{different}) = \frac{40+40}{1000}=0.08.
$$

Therefore,

$$
E = 0.92-0.08=0.84.
$$

The value of $E$ ranges from $-1$ to $+1$:

* $E=+1$: perfectly correlated, always the same
* $E=-1$: perfectly anti-correlated, always different
* $E=0$: no correlation

## 3. Calculate the CHSH value

Calculate the four correlators:

$$
E(a,b),\quad E(a,b'),\quad E(a',b),\quad E(a',b').
$$

Then combine them as

$$
S =
E(a,b)
-
E(a,b')
+
E(a',b)
+
E(a',b').
$$

For the Bell state and the measurement angles above, you should obtain a value **close to**

$$
S=2\sqrt{2}\approx2.83,
$$

with small deviations due to the finite number of shots.
