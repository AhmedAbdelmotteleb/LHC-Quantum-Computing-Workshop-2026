# CHSH / Bell Inequality Exercise

## Context & Motivation

In Lecture 2 you proved (slides 16–17) that a Bell state cannot be written as a product of two single-qubit states. The two qubits are entangled. This exercise makes that fact *measurable*: you'll compute a quantity called the **CHSH value (S)** from real measurement statistics, and show that entangled qubits produce correlations that are impossible for any classical ("local hidden variable") system to reproduce.

**The classical bound:** for any local hidden-variable theory, $S \le 2$.
**The quantum bound (Tsirelson's bound):** entangled qubits can reach $S = 2\sqrt{2} \approx 2.828$.

Seeing S exceed 2 in your own simulation is about as close as a workshop gets to personally disproving local realism.

## Problem Statement

Alice and Bob each hold one qubit of a shared Bell state $|\Phi^+\rangle = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$. Each of them independently picks one of **two possible measurement settings**:

- Alice measures at angle $a=0$ or $a'=\pi/2$
- Bob measures at angle $b=\pi/4$ or $b'=3\pi/4$

(These angles are on the Bloch sphere; you can implement "measuring at angle $\theta$" by applying `RY(-theta)` before a standard Z-basis measurement.)

For each of the four setting combinations $(a,b), (a,b'), (a',b), (a',b')$, you'll compute the **correlator**:
$$E(\theta_a,\theta_b) = P(\text{same outcome}) - P(\text{different outcome})$$

Then combine them into the CHSH quantity:
$$S = E(a,b) - E(a,b') + E(a',b) + E(a',b')$$
