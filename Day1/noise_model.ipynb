## Context & Motivation

In Q5 you saw that **amplitude damping** noise biases a Hadamard-then-measure circuit toward $|0\rangle$. But amplitude damping is only one of several distinct physical noise processes real qubits suffer from. This exercise has you compare three:

- **Amplitude damping** : energy relaxation (T1 decay): $|1\rangle \to |0\rangle$ over time. Asymmetric: it only pushes toward the ground state.
- **Phase damping** : pure dephasing (T2 decay): randomizes the *relative phase* between $|0\rangle$ and $|1\rangle$ without changing the probability of measuring either one. Doesn't move population, only scrambles coherence.
- **Depolarizing noise** : with some probability, the qubit's state is replaced entirely by the maximally mixed state. Symmetric: doesn't favor $|0\rangle$ or $|1\rangle$.

The interesting question: **which of these three actually changes your measurement histogram, and which don't?**
